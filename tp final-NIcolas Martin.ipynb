**1. Porcentaje Obtenido** (Sobre Votos Válidos o Afirmativos)
Este es el porcentaje que se anuncia más comúnmente en los medios y determina quién gana la elección.

Denominador (Base 100%): El total de votos válidos o, más específicamente, el total de votos afirmativos para una categoría de cargo (excluyendo votos en blanco y nulos).

Qué Mide: Mide la preferencia o el ***respaldo de un partido entre aquellos ciudadanos que efectivamente votaron*** y lo hicieron de forma válida.

**2. Proporción sobre el Padrón** (Sobre Electores Habilitados)
Esta proporción se utiliza para medir el alcance y el impacto real de un partido dentro del total de la población habilitada para votar.

Denominador (Base 100%): El Padrón Electoral (el número total de ciudadanos habilitados para votar en esa jurisdicción, sin importar si votaron o no).

Qué Mide: Mide el respaldo potencial o la** fuerza estructural de un partido en relación con todo el universo de electores, incluidos los que se abstuvieron.**
